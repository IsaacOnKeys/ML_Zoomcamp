```python
asian_cars = df[df.origin == "Asia"]
```

```python
weight_year = asian_cars[["vehicle_weight", "model_year"]]
```

```python
X = np.array(weight_year.head(n=7))
```

```python
XTX = X.T @ X
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
y
```

```python
w = (XTX_inv @ X.T) @ y
w.sum()
```
